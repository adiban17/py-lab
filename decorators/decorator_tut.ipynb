### Decorators

```python
import time
```

```python
def brew_tea():
    start = time.perf_counter()
    print("Brewing Tea...")
    time.sleep(1)
    print("Tea is ready")
    end = time.perf_counter()
    print(f"Time taken to run:{end-start} seconds")

brew_tea()
    
```

```python
def make_matcha():
    start = time.perf_counter()
    print("Making Matcha...")
    time.sleep(1)
    print("Matcha is ready")
    end = time.perf_counter()
    print(f"Time taken to run:{end-start} seconds")

make_matcha()
    
```

In the above cells, we are trying to time our function, but we are rewriting our timing code. This is not optimal. Thus we use decorators to add additional functionality of timekeeping to our `brew_tea()` and `make_matcha()` functions.

Creating a Decorator Function

```python
def timer_dec(base_fn):
    def enhanced_fn():
        start = time.perf_counter()
        base_fn()
        end = time.perf_counter()
        print(f"Time taken to run : {end-start} seconds")
    return enhanced_fn
```

Adding the decorator to our functions

```python
@timer_dec
def brew_tea():
    print("Brewing Tea...")
    time.sleep(1)
    print("Tea is ready")
    
brew_tea()
```

```python
@timer_dec
def make_matcha():
    print("Making Matcha...")
    time.sleep(1)
    print("Matcha is ready")

make_matcha()
```

The above examples were functions without params. Now we will try running functions with arguments

```python
@timer_dec
def brew_bev(bev_name, steep_time):
    print(f"Brewing {bev_name}.....")
    time.sleep(steep_time)
    print(f"{bev_name} is ready!")

brew_bev("Tea", 1)
```

The above error shows that we need to make our decorator immune to passing arguments.

Redefining our decorator

```python
def timer_dec(base_fn):
    def enhanced_fn(*args, **kwargs):
        start = time.perf_counter()
        base_fn(*args, **kwargs)
        end = time.perf_counter()
        print(f"Time taken to run : {end-start} seconds")
    return enhanced_fn
```

Testing it out

```python
@timer_dec
def brew_bev(bev_name, steep_time):
    print(f"Brewing {bev_name}.....")
    time.sleep(steep_time)
    print(f"{bev_name} is ready!")

brew_bev("Tea", 2)
```

```python
@timer_dec
def brew_bev(bev_name, steep_time):
    print(f"Brewing {bev_name}.....")
    time.sleep(steep_time)
    print(f"{bev_name} is ready!")

brew_bev(bev_name="Matcha", steep_time=5)
```

Yaaaaayyyy!!!

- `*args` handles arguments
- `**kwargs` handles keyword arguments
