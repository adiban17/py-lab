## Memoization

Imagine running a function with the same arguments twice. 
```
calc_square(2)
calc_square(2)
```
If it takes `x` seconds to run function `calc_square()`. Then it will take `2x` time to run it twice.
Instead, we store the results in cache so that if the same result is called twice. Then we can get the results quickly without recalculating the outputs.
This process is called memoization.

Imports

````python
import time
from functools import lru_cache
````

### Memoization Demo (Manual)

````python
def slow_square(n):
    time.sleep(2)
    result = n **2
    return result

start = time.time()
slow_square(2)
slow_square(2)
end = time.time()
print(round(end-start, 2))
````

We run the above code without memoization. Makes it take more time to calculae the square of 2. Even though the result has been previously calculated earlier.

````python
square_cache = {}

def slow_square(n):
    if n in square_cache:
        return square_cache[n]
    time.sleep(2)
    result = n **2
    square_cache[n] = result
    return result

start = time.time()
slow_square(2)
slow_square(2)
end = time.time()
print(round(end-start, 2))
````

We can see that in this run, the code is running fatser as we use the `square_cache` dict which already has the results of the square of 2. Fetching it faster rather than calculating it again and again.

But a problem is that the above code is pretty messy. We need to make the code cleaner and more reusable and more readable.

### Memoization (using `lru_cache`)

````python
@lru_cache
def slow_square(n):
    time.sleep(2)
    result = n **2
    return result

start = time.time()
slow_square(2)
slow_square(2)
end = time.time()
print(round(end-start, 2))
````

````python
@lru_cache
def slow_square(n):
    time.sleep(2)
    result = n **2
    return result

slow_square(2)
slow_square(2)
slow_square(3)
slow_square(4)
slow_square(3)

slow_square.cache_info()
````

In the above cells, we use `@lru_cache` which is a decorator which helps us use cache memory without having to explicitly define a `square_cache` dict.

- `hits`: tells us how many time the cache memory was used during function calling
- `misses`: tells us the number of times cache memory could not be used during function calling
- `currsize`: tells us the number of function results stored in the cache
- `maxsize` : the maximum number of results that we can store in the cache

Keep in mind that the `@lru_cache` is not free. It takes up system memory. If we use cache for a function whihc does not take a lot of time run then there is no need for us to use cached memory. Use cache only if time savings is more than time cost and storage cost.

Another thing to keep in mind is that, we should be using cache memory only for pure functions. 
By pure functions we mean functions whose output remains the same for the same input. For example: if we pass 2 as an argument then the result should be 4 every time we pas 2, if the result is sometimes 4 and sometimes 8 then it is not a pure function and it if we do use cached memory then we will be getting the result as 4 even if the expected output is 8, becasue 4 was the value that was initially stored in the cache.

Do not use memoization in the following cases where your function is dependent on the following:
- Reads from Database
- Uses current time
- Returns random #

We should also not cache when the items that we need to store as the key in the cached dict are the following (since they are unhashable types):
- lists
- dictionaries
- sets

The following are ideal for cached dict keys:
- strings
- numbers
- tuples
- frozensets

`@lru_cache` has a maximum size of 128 elements. If anythong more than 128 elements are being stored in the cache then the least recently used element is thrown away. `lru` in `@lru_cache` stands for `least recently used`.

If in case we do want to store more elements in the cache. Then we can pass the following argument : `@lru_cache(maxsize=None)`. Or we can also use `@cache` as the cache decorator. The bigger the cache size we have, the more time it takes for the cache to fetch the value from the memory.
