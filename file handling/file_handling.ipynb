### File Handling

Opening a file and reading it

```python
file = open("shopping_list.txt", "r")
contents = file.read()
file.close()
print(contents)
```

Showing the raw representation of a file

```python
print(repr(contents))
```

Splitting each item on a new line as an item in a list

```python
items = contents.split("\n")
items.sort()    # sorting the items in alphabetical order

for item in items:
    print(" - " + item)
```

Writing Data to a File

```python
items = "brocolli\nrice\nchicken\ncheese"
file = open("groceries.txt", "w")
file.write(items)
file.close()
```

Appending Items to a File

```python
file = open("groceries.txt", "a")
items = "\nyogurt"
file.write(items)
file.close()
```

Context Managers: 

- The `close()` function is very important in the whole code here. If in case we do not close the file after opening it, then the changes we make to the file will not be saved. This is beacuse the changes that we make to the file are stored in a memory buffer before they are written to the actual file. 

- This done as writing data to files is very slow, and thus Python stores the immediates changes in memory buffer and batched those changes before writing it to the actual file.If we do not close the actual file then the changes will be lost. 

- Moreover opening multiple files and never closing them will take up a lot of computational resouces. Thus in order to not let that happen, we use something called as context manager.

```python
# while using a context manager we do not need to use `close()`
with open(
    "groceries.txt" #here we put the file path,
     "a") as file:
    file.write("\nbananas")
```

Different Modes in which we can opne files:
- `r` - read mode
- `w` - write mode
- `a` - append mode
- `x` - exclusive crete (raises error if the file exists)
- `r+` - read + write
- `w+` - write + read
- `a+` - append + read
- `x+` - exclusive create + read
- `rb`,`wb`,`ab`,`rb+`,`wb+`,`ab+` - read, write, append binary (all these modes are used for non-textual formats such as images, videos, PDFs) to prevent data corruption
