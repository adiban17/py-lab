```python
import json
```

Reading a JSON file

```python
with open('catalog.json') as f:
    catalog = json.load(f)
    
catalog
```

Printing each item in the json objects

```python
for item in catalog["catalog"]:
    print(f"Name: {item["name"]}")
    print(f"Name: {item["price"]}")
    print(f"Name: {item["prep time"]}")
    print()
```

Adding a new drink to the JSON Catalog and then updating the catalog

```python
with open('catalog.json', 'r+', encoding='utf-8') as f:
    catalog = json.load(f)

catalog
```

```python
catalog["catalog"].append({
    "name":"Boba Tea",
    "price":250,
    "discounted price":200,
    "prep time":3
})
```

```python
catalog["catalog"]
```

```python
with open('updated_catalog.json', 'w') as f:
    json.dump(catalog, f, indent=2)
```
